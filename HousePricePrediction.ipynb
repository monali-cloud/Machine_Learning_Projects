```python
#import Libraries
import pandas as pd
import matplotlib.pyplot as plt

from sklearn.linear_model import LinearRegression

print("ALL libraries are imported successfully")
```

```python
#Create Dataset 
data= {
    "Area":[500,700,900,1100,1300,1500,1700,1900,2100,2300],
    "Price":[15,20,26,32,38,45,50,57,63,70]
}

```

```python
df = pd.DataFrame(data)
df
```

```python
X= df[["Area"]]
Y= df["Price"]
```

```python
model = LinearRegression()

model.fit(X,Y)
```

```python
Area=[[1800]]

predict= model.predict(Area)

print("Predicted House Area Price is :",predict[0],"Lakhs")
```

```python
plt.scatter(df["Area"],df["Price"])

plt.xlabel("Area")
plt.ylabel("Price")
plt.title("House Price Prediction")

plt.show()
```

```python
plt.figure(figsize=(8,5))

plt.scatter(
    df["Area"],
    df["Price"],
    color="blue",
    s=100,
    label="Actual Data"
)

plt.plot(
    df["Area"],
    model.predict(X),
    color="red",
    label="Regression Line"
)

plt.xlabel("Area (sq.ft)")
plt.ylabel("Price (Lakhs)")
plt.title("House Price Prediction")

plt.legend()
plt.grid(True)

plt.show()
```
